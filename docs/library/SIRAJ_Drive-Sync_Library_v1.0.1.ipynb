## ١ ـ تسجيل الدخول إلى Google Drive

```python
from google.colab import auth
auth.authenticate_user()
from googleapiclient.discovery import build
drive = build('drive', 'v3')
print('✓ Drive جاهز:', drive.about().get(fields='user(emailAddress)').execute()['user']['emailAddress'])
```

## ٢ ـ تنزيل المكتبة من GitHub
إن كان المستودع خاصًا، الصق رمز GitHub (صلاحية قراءة فقط تكفي). الرمز لا يُحفظ في الدفتر.

```python
import getpass, os, subprocess
token = getpass.getpass('رمز GitHub (اتركه فارغًا إن كان المستودع عامًا): ').strip()
url = f'https://{token}@github.com/Mr-Ghassan/siraj.git' if token else 'https://github.com/Mr-Ghassan/siraj.git'
if not os.path.isdir('/content/siraj'):
    r = subprocess.run(['git', 'clone', '--depth', '1', '--branch', 'feat/siraj-brand-library-completion-v1', url, '/content/siraj'], capture_output=True, text=True)
    print(r.stderr.replace(token, '***') if token else r.stderr)
    subprocess.run(['git', '-C', '/content/siraj', 'remote', 'set-url', 'origin', 'https://github.com/Mr-Ghassan/siraj.git'])
del token, url
print(subprocess.run(['git', '-C', '/content/siraj', 'log', '-1', '--format=%h %s'], capture_output=True, text=True).stdout)
```
